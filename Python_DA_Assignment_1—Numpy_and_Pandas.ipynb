###Numpy Array Operations:

```python
#Tasks:
#1. Create a 1D NumPy Array:
import numpy as np
#1 Dimentional array
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week1 Temp:",temperatures_w1)
```

```python
#2. Inspection and Properties:
#shape, data type, and the number of elements in the array
print("Shape:", temperatures_w1.shape)
print("Data Type:",temperatures_w1.dtype)
print("Number of Elements:",temperatures_w1.size)
```

```python
#3. Array Operations:
#Convert the temperatures_w1 from Celsius to Fahrenheit
#Fahrenheit = (Celsius * 9/5) + 32
Fahrenheit=(temperatures_w1*9/5) + 32
print("Temp in Fahrenheit:",Fahrenheit)
```

```python
# Find the maximum, minimum, and mean temperatures for the week.
print("Max Temperature:", np.max(temperatures_w1))
print("Min Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
#4. Array Slicing and Indexing:
#● Extract the temperatures for the first three days of the week.
print("First 3 Days:", temperatures_w1[:3])
```

```python
#● Extract the temperatures for the weekend (last two days).
print("Last 2 days:",temperatures_w1[-2:])
```

```python
#● Extract the temperatures for the middle three days of the week.
print("Middle 3 days:",temperatures_w1[2:5])
```

```python
#5. Create a 2D Array:

#Create a 2D NumPy array named temperatures where each row represents
#temperatures for a different week:
#★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
#★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)
temperature=np.array([[22.5,25.3,20.8,23.4,26.1,24.8,21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(f"Two weeks Temperatures:")
print(temperature)
```

```python
# 6.Inspect and Slice the 2D Array:
# shape, data type, and total number of elements
print("Shape:", temperature.shape)
print("Data Type:", temperature.dtype)
print("Total Elements:", temperature.size)
```

```python
#Extract the temperatures for each week and the weekends (last two days) for
#both weeks.

print("Week 1:", temperature[0])  # Week 1


print("Week 2:", temperature[1])   # Week 2


print("Week 1 Weekend:", temperature[0, -2:])# Weekend temperatures for both weeks
print("Week 2 Weekend:", temperature[1, -2:])
```

##Pandas Series:


```python
#1. Creating Pandas Series:
import pandas as pd
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
    )
marks
```

```python
#2. Indexing and Slicing:
#● Use the integer index position to access the mark of the 1st rank student.
print("1st Rank Mark:", marks.iloc[0])

#● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index
#labels.
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

#● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
print("3rd Rank Mark:", marks.iloc[2])

#● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are
#greater than 90.
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Series:
#● Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1'] = 100
print(marks)

```

```python
#● Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')
print(marks)
```

```python
#● Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks / 10
print(f"CGPA")
print(cgpa)
```

##Pandas DataFrame:

###1. Creating Pandas DataFrame:

```python
import pandas as pd

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
##2. Data Exploration:
##● Display the transactions DataFrame and its basic information, including head, tail,
##shape, column names, and data types.
print(transactions.shape)
print(transactions.dtypes)
print(transactions.info())
print(transactions.describe())
```

```python
#● Display only the 'ProductCategory' and 'Amount' columns.

print(transactions[['ProductCategory', 'Amount']])
```

```python
#● Retrieve the last 3 columns of the DataFrame using iloc or loc.
print(transactions.iloc[:, -3:])
```

```python
#● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(result)
```

```python
#● Find the value counts for the 'ProductCategory' column.
print(transactions['ProductCategory'].value_counts())
```

```python
#● Find the unique values in the 'Region' column.
print(transactions['Region'].unique())
```

```python
#● Group by 'Region' and find the mean amount for each region.
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#3. Manipulating the DataFrame:
#● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print(transactions)
```

```python
#● Add a new column 'Discount' by calculating 10% of the 'Amount'.
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
#● Remove the row with TransactionID 109.
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

```python
#● Delete the 'Discount' column from the DataFrame.
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```
