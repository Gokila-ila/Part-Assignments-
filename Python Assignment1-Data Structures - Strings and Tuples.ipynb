#### Tuples (Creation, Modification and Access)

```python
#Create the 1st tuple with values -> (10, 20, 30) and the 2nd tuple with values -> (40, 50, 60):
tuple1=(10,20,30)
tuple2=(40,50,60)
print("Tuple1 :",tuple1)
print("Tuple2 :",tuple2)
```

```python
## a. Concatenate the two tuples and store it in “t_combine”
t_combine=tuple1+tuple2
print("t_combine =",t_combine)
```

```python
## b. Repeat the elements of “t_combine” 3 times
print("Repeated Elements : ",t_combine*3)

```

```python
# c. Access the 3rd element from “t_combine”
print("3rd Element:",t_combine[2])

```

```python
## d. Access the first three elements from “t_combine”
print("First 3 Elements :",t_combine[0:3])
```

```python
## e. Access the last three elements from “t_combine”
print("Last 3 Elements :",t_combine[-3:])
```
