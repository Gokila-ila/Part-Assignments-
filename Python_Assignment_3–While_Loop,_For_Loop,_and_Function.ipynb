###Task 1: While Loop & Control Statements -Number Guessing Game

```python
#Task:1.
#Generate random number between 1 and 10
import random
secret_number = random.randint(1,10)
attempts=3
while attempts>0:
   guess=int(input(f"Guess the number between 1 & 10 "))
   if guess < 1 or guess > 10:
      print(f"Your guessing number is out of range, Please guess a number between 1 to 10: ")
      continue
   attempts-=1
   if guess==secret_number:
      print(f"Congratulations you guessed the correct number")
      break
   elif  guess> secret_number:
      print(f"Too high. Try again")
   else:
      print(f"Too low. Try again")
else:
   print(f"Better luck next time.")
   print(f"The correct number is  ",secret_number)
```

###Task 2:- For Loop - Multiplication Table Generator

```python
# Program for multiplication table:
number=int(input("Enter the number for Multiplication  table you want"))
for i in range(1,11):
   print(f"{number} * {i} ={number*i}")

```

###Task 3: Function- BMI Calculator

```python
#Program for BMI Calculator
def calculate_bmi(weight,height):
   bmi=weight/(height**2)
   return bmi

weight=float(input("Enter your weight in kg:  "))
height=float(input("Enter your height in meters: "))
bmi = calculate_bmi(weight,height)
print("Your BMI is: ",bmi)
```
